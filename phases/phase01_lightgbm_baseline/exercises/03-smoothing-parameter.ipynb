## Solution

```python
thresholds = [0.3, 0.5, 0.7]

for threshold in thresholds:
    calibrator = IsotonicCalibrator(threshold=threshold)
    calibrator.fit(raw_probs, y_test)
    cal_probs = calibrator.predict(raw_probs)
    
    brier = brier_score_loss(y_test, cal_probs)
    print(f"Threshold {threshold}: Brier score = {brier:.4f}")
```

**Observation:** The threshold parameter affects the decision boundary but not necessarily the calibration quality. Lower Brier scores indicate better calibration.
